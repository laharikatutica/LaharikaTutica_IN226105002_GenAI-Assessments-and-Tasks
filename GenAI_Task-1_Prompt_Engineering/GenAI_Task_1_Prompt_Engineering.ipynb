### **Task 1: Replace Hardcoded Prompts**

```python
# PromptTemplate for explaining a topic
explain_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)

# Reusable function using the template (no f-strings, no hardcoding)
def explain_topic(topic):

    return explain_template.format(topic=topic)

# Testing the template with different topics
print("Test case: 1")
print(explain_topic("Machine Learning"))

print("\nTest case: 2")
print(explain_topic("Neural Networks"))

print("\nTest case: 3")
print(explain_topic("Python"))
```

### **Task 2: Multi-Input Prompt System**

### **Task 4: ChatPromptTemplate System**

```python
# Define role descriptions
role_descriptions = {
    "teacher":     "You are an expert teacher. Explain concepts clearly with examples and step-by-step guidance.",
    "interviewer": "You are a professional interviewer. Ask insightful, thought-provoking questions about the topic.",
    "motivator":   "You are an enthusiastic motivator. Inspire the user to learn and explore the topic with energy."
}

def build_chat_prompt(role, topic):

    # Check if role is valid
    if role not in role_descriptions:
        raise ValueError("Role must be one of: teacher, interviewer, motivator")

    # Create system and user message templates
    system_message = SystemMessagePromptTemplate.from_template("{role_desc}")

    human_templates = {
    "teacher": "Explain {topic} to me as a student would understand.",
    "interviewer": "Ask me insightful questions about {topic}.",
    "motivator": "Encourage me to learn about {topic} enthusiastically!"
    }

    human_message = HumanMessagePromptTemplate.from_template(human_templates[role])

    # Combine into chat prompt
    chat_prompt = ChatPromptTemplate.from_messages([system_message, human_message])

    # Format messages with actual role description and topic
    formatted_messages = chat_prompt.format_messages(
        role_desc=role_descriptions[role],
        topic=topic
    )
    return formatted_messages

def display_chat_prompt(role, topic):
    messages = build_chat_prompt(role, topic)
    print("[ROLE: {} | TOPIC: {}]".format(role, topic))
    for msg in messages:
        print("  {}: {}".format(msg.type.upper(), msg.content))

# --- Test Cases ---
test_cases = [
    {"role": "teacher", "topic": "Neural Networks"},
    {"role": "interviewer", "topic": "Deep Learning"},
    {"role": "motivator", "topic": "Artificial Intelligence"}
]

for i, case in enumerate(test_cases):
    print(f"\nTest case {i+1}:")
    display_chat_prompt(case['role'], case['topic'])
```

### **Task 5: Input Validation Layer**

```python
def validate_inputs(audience, tone, raise_error=False):
    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    warnings = []

    if audience not in valid_audience:
        msg = "Invalid audience '{}'. Must be one of: {}".format(audience, valid_audience)
        if raise_error:
            raise ValueError(msg)
        else:
            warnings.append("[WARNING] {}. Using default: 'beginner'".format(msg))
            audience = "beginner"

    if tone not in valid_tone:
        msg = "Invalid tone '{}'. Must be one of: {}".format(tone, valid_tone)
        if raise_error:
            raise ValueError(msg)
        else:
            warnings.append("[WARNING] {}. Using default: 'casual'".format(msg))
            tone = "casual"

    return audience, tone, warnings

def test_validate(audience, tone, raise_error=False):
    print("Audience: {} | Tone: {}".format(audience, tone))
    try:
        validated_audience, validated_tone, warnings = validate_inputs(audience, tone, raise_error)
        for w in warnings:
            print("  {}".format(w))
        print("Validated: audience={}, tone={}".format(validated_audience, validated_tone))
    except ValueError as e:
        print("Caught ValueError: {}".format(e))

# --- Test Cases ---
print("Test Case 1: Valid inputs")
test_validate("beginner", "fun")

print("\nTest Case 2: Invalid inputs (defaults applied)")
test_validate("child", "dramatic")

print("\nTest Case 3: Invalid inputs (error raised)")
test_validate("child", "dramatic", raise_error=True)
```

### **Task 6: Prompt Generator App**

```python
def generate_prompt_report(topic, audience, tone, style):

    audience, tone, warnings = validate_inputs(audience, tone)

    style_templates = {
        "teaching": "Explain {topic} for {audience} in a {tone} style",
        "interview": "Ask 3 questions about {topic} for {audience} in a {tone} tone",
        "storytelling": "Explain {topic} for {audience} in a {tone} storytelling style"
    }

    if style not in style_templates:
        raise ValueError("Invalid style. Choose: teaching, interview, storytelling")

    prompt_template = PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template=style_templates[style]
    )

    prompt_text = prompt_template.format(topic=topic, audience=audience, tone=tone)

    for w in warnings:
        print("{}".format(w))
    print("{}".format(prompt_text))

# --- Test Cases ---
test_cases_task6 = [
    {"topic": "Neural Networks", "audience": "beginner", "tone": "fun", "style": "storytelling"},
    {"topic": "Python", "audience": "expert", "tone": "formal", "style": "interview"},
    {"topic": "Deep Learning", "audience": "intermediate", "tone": "casual", "style": "teaching"}
]

for i, case in enumerate(test_cases_task6):
    print(f"\nTest Case {i+1}:")
    generate_prompt_report(**case)
```

### **Task 7: Template Reusability Test**

```python
# Single reusable master template
master_template = PromptTemplate(
    input_variables=["topic", "audience", "tone", "style"],
    template=(
        "You are a {style}. "
        "Explain {topic} to a {audience}-level learner in a {tone} tone."
    )
)

# 5 completely different input combinations
reusability_inputs = [
    {"topic": "Machine Learning",   "audience": "beginner",     "tone": "fun",     "style": "teacher"},
    {"topic": "Deep Learning",         "audience": "intermediate", "tone": "casual",  "style": "motivator"},
    {"topic": "Artificial Intelligence",  "audience": "expert",       "tone": "formal",  "style": "researcher"},
    {"topic": "Natural Language Processing", "audience": "beginner", "tone": "friendly", "style": "trainer"},
    {"topic": "Reinforcement Learning", "audience": "intermediate", "tone": "technical", "style": "interviewer"},
]

for i, inputs in enumerate(reusability_inputs, start=1):
    # Same template, different data each time
    output = master_template.format(**inputs)
    print("\nTest case {}:".format(i))
    # Display the inputs used for each run
    print("Input : topic={}, audience={}, tone={}, style={}".format(
        inputs['topic'], inputs['audience'], inputs['tone'], inputs['style']
    ))
    # Display the generated output
    print("Output: {}".format(output))
```

### **Pipeline Workflow Code**

```python
def validate_inputs(audience, tone, raise_error=False):
    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    warnings = []

    if audience not in valid_audience:
        msg = "Invalid audience '{}'. Must be one of: {}".format(audience, valid_audience)
        if raise_error:
            raise ValueError(msg)
        else:
            warnings.append("[WARNING] {}. Using default: 'beginner'".format(msg))
            audience = "beginner"

    if tone not in valid_tone:
        msg = "Invalid tone '{}'. Must be one of: {}".format(tone, valid_tone)
        if raise_error:
            raise ValueError(msg)
        else:
            warnings.append("[WARNING] {}. Using default: 'casual'".format(msg))
            tone = "casual"

    return audience, tone, warnings

def full_pipeline(topic, audience, tone, style):

    print("=" * 50)
    print("MINI PROMPT ENGINE – Full Pipeline")
    print("=" * 50)

    # Step 1: Display initial user inputs for clarity.
    print("Step 1: User Input   → topic={}, audience={}, tone={}, style={}".format(topic, audience, tone, style))

    # Step 2: Validate audience and tone. Invalid inputs default to 'beginner' and 'casual'.
    validated_audience, validated_tone, warnings = validate_inputs(audience, tone, raise_error=False)
    for w in warnings:
        print("  {}".format(w))
    print("Step 2: Validation   → audience={}, tone={}".format(validated_audience, validated_tone))

    # Step 3: Template selection and formatting - Generate the final prompt string.
    print("Step 3: Template     → Style '{}' template selected and formatted".format(style))

    style_templates = {
        "teaching": "Explain {topic} for {audience} in a {tone} style",
        "interview": "Ask 3 questions about {topic} for {audience} in a {tone} tone",
        "storytelling": "Explain {topic} for {audience} in a {tone} storytelling style"
    }

    if style not in style_templates:
        raise ValueError("Invalid style. Choose: teaching, interview, storytelling")

    prompt_template = PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template=style_templates[style]
    )

    final_prompt = prompt_template.format(topic=topic, audience=validated_audience, tone=validated_tone)

    # Step 4: Display the final generated prompt.
    print("Step 4: Final Output → {}".format(final_prompt))
    print("=" * 100)
    return final_prompt

# Run the full pipeline with an example to demonstrate its functionality.
full_pipeline(
    topic="Artificial Intelligence",
    audience="beginner",
    tone="fun",
    style="storytelling"
)
```

**Pipeline Flow**

User Input → Validation → Prompt Template → Dynamic Prompt Generation → Output
