## **1. Introduction to LangChain**
**1.1 What is LangChain?**

LangChain is an open-source framework that helps developers build applications powered by Large Language Models (LLMs) using modular and reusable components. Instead of directly calling APIs like OpenAI or HuggingFace, LangChain acts as a middleware layer that connects models, data, and tools into a structured pipeline.
In simple terms, LangChain acts as a bridge between raw LLM capabilities and real-world applications.

**1.2 Why is LangChain Important?**

Modern AI applications are not just about generating text — they require:
- Context awareness
- External data integration
- Multi-step reasoning
- Tool usage (APIs, databases, etc.)

Without LangChain, developers must manually manage:
- Prompt engineering
- State handling
- API orchestration
- Tool integration
LangChain abstracts all of this into reusable components.

**1.3 Problems LangChain Solves**

1. Orchestration Problem

Managing multiple LLM calls and logic becomes messy. LangChain provides structured pipelines.

2. Chaining Logic

Instead of one prompt → one response, you can build multi-step workflows.

3. Tool Integration

LLMs can interact with:
- APIs
- Databases
- Search engines

4. Memory Handling

Maintains conversation history and context.

## **Core Components**

#### **1. LLMs**

**Concept:**

At the core of LangChain is the language model interface, which connects to providers like OpenAI or HuggingFace. LangChain standardizes LLM calls so developers can switch models without rewriting logic.

**Why it exists:**

- Abstracts different LLM APIs into a unified interface
- Enables model switching (GPT → Claude → LLaMA)
- Supports chat-based and completion-based models

```python
# Loading lightweight free model
pipe = pipeline(
    "text-generation",
    model="mistralai/Mistral-7B-Instruct-v0.1",
    max_new_tokens=100
)

llm = HuggingFacePipeline(pipeline=pipe)
```

```python
print(llm.invoke("What is LangChain in AI?"))
```

#### **2. Prompt Templates**

**Concept**

Prompts are instructions given to the LLM. Prompt Templates are dynamic templates with placeholders that can be filled at runtime.

**Why it exists**

Hardcoding prompts is inefficient, error-prone, and not reusable. Templates allow developers to structure prompts consistently, pass dynamic variables, and manage complex instructions at scale. They also improve maintainability of multi-step LLM workflows.

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    "Explain {topic} in simple terms"
)

print(prompt.format(topic="LangChain"))
```

#### **3. Chains**

**Concept**

Chains combine multiple components into a pipeline. LangChain now offers LCEL (LangChain Expression Language) for more advanced chaining.

**Why it exists**

Chains orchestrate multi-step reasoning and allow outputs from one component to feed another.

```python
from langchain_core.output_parsers import StrOutputParser

chain = prompt | llm | StrOutputParser()

print(chain.invoke({"topic": "LangChain Chains"}))
```

#### **4. Memory**

**Concept**

Memory allows LangChain applications to retain past interactions, enabling context-aware responses. For example, remembering a user’s name across multiple queries.

**Why it exists**

LangChain provides multiple memory types like:
- Buffer Memory (stores full conversation)
- Token Memory (limits by tokens)
- Summary Memory (stores summarized history)

```python
#Import Libraries
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain.memory import ConversationBufferMemory
```

```python
#Memory Example
print("Memory Example:")

memory = ConversationBufferMemory()

memory.save_context({"input": "Hello"}, {"output": "Hi there!"})
memory.save_context({"input": "What is AI?"}, {"output": "AI stands for Artificial Intelligence"})

print(memory.load_memory_variables({}))
```

#### **5. Tools**

**Concept**

Tools are external functions that LLMs can call (APIs, calculators, search engines, etc.)

**Why it exists:**

- LLMs alone cannot access external systems
- Tools extend capabilities beyond text generation

```python
from langchain.tools import tool

@tool
def calculator(expr: str) -> str:
    """Evaluates a mathematical expression and returns the result."""
    return str(eval(expr))

print(calculator.run("10 * 5"))
```

```python
question = "What is 20 * 4?"
print(chain.invoke({"topic": question}))
```

#### **6. Agents with tools**

**Concept**

Agents are decision-making systems that dynamically choose:
- which tool to use
- what action to take

Unlike chains (fixed flow), agents are dynamic and reasoning-based.

```python
# Agent
from langchain.agents.agent import initialize_agent, AgentType

agent = initialize_agent(
    tools=[calculator],
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    handle_parsing_errors=True,
    verbose=True
)

print(agent.run("What is 12 * 8 + 5?"))
```

#### **7. Document Loader**

**Concept**

Document loaders help ingest external data like PDFs, text files, or web pages into LangChain.

**Why it exists:**

- LLMs need external knowledge
- Enables RAG (Retrieval Augmented Generation)

```python
from langchain_community.document_loaders import TextLoader

with open("example.txt", "w") as f:
    f.write("LangChain is an open-source framework that helps developers build applications powered by Large Language Models (LLMs) using modular and reusable components.")

loader = TextLoader("example.txt")
docs = loader.load()

print(docs)
```

## **6. Advantages and Limitations**
**Advantages**
- Modular design → LangChain is built with separate components (prompts, chains, tools), making it easy to mix and reuse parts.
- Easy prototyping → Quickly build LLM apps without writing complex backend logic from scratch.
- Strong integrations → Works smoothly with OpenAI, HuggingFace, APIs, databases, and external tools.
- Supports RAG pipelines → Enables retrieval-augmented generation for answering from custom documents.
- Scalable architecture → Can be extended from simple chatbots to complex multi-agent systems.

**Limitations**
- Latency (multi-step chains) → Multiple LLM calls in a pipeline can slow down response time.
- Debugging complexity → Hard to trace errors across chains, agents, and tool calls.
- Cost (API usage) → More components = more LLM calls, increasing token costs.
- Overkill for simple tasks → For basic prompts, LangChain adds unnecessary complexity compared to direct API calls.

**When NOT to Use LangChain**
- Simple one-shot LLM tasks
- Small scripts
- Low-latency critical systems
